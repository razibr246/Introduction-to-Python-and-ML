# array creation with random values

```python
# np.random.randint(start,end, shape)

arr = np.random.randint(1,10 , (2,3))

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

array creation with range functions

```python
# np.arange(first , end , step ) -> similar to range function

arr = np.arange(1 , 10 ,1 )
mat = arr.reshape(3,3)

print(arr)
print(mat)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.linspace(start , end(included) , koita)

arr = np.linspace(0, 4 ,6)


print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.logspace(start , end(included) , koita )

arr = np.logspace(0 , 4 ,6, base =2)

print(arr)
```

```python
import numpy as np

# create a 2D array
arr2 = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

print("Original array:")
print(arr2)


# --------------------
# getting rows
# --------------------
row_0 = arr2[0:1, :]   # first row (2D)
row_1 = arr2[1:, :]    # from second row onward

print("\nFirst row:")
print(row_0)

print("\nRemaining rows:")
print(row_1)


# --------------------
# getting columns
# --------------------
col_0 = arr2[:, 0]     # first column (1D)
col_1 = arr2[:, 1]     # second column (1D)

print("\nFirst column:")
print(col_0)

print("\nSecond column:")
print(col_1)


# --------------------
# getting a portion
# --------------------
portion = arr2[:, 0:2]  # first two columns of all rows

print("\nPortion (first 2 columns):")
print(portion)
```
