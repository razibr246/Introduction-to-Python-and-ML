### Splitting,Joining, Formatted String

```python
prompt = "What is Phitron ?"

tokens = prompt.split()

print(tokens)
```

```python
prompt = "What ;is ;Phitron - ?"

tokens = prompt.split(";")

print(tokens)
```

```python
prompt = "What is Phitron ?"

tokens = prompt.split()

print(tokens)
print(type(tokens))
print(type(tokens[0]))

sentence = "-".join(tokens)

print(sentence)
print(type(sentence))
```
