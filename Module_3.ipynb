### string Method

```python
string = "Welcome to Phitron ML Course,Mr.Phitron. To Be ML Enginner, ML Expert To start a career"

#checking photron lower case in matched or not!
print("Just checking the phitron lower case was in there . it will print false .")
print("phitron" in string)
processed_string = string.lower()
#simple printing the lower case letter worlds
print(processed_string)
#Length

print(len(string))
#The Upper case Phitron is found
print("Phitron" in string )
print("ML" in string )

#checking only the "phitron" word is there or not . it will return as True!
print( "phitron" in processed_string)

#finding sbstring in the sentence

start_index = processed_string.find("phitron")
last_index = processed_string.rfind("phitron")

print(start_index)
print(last_index)

#substrin how many times!

count = processed_string.count("phitron")

print(count)


#replacing the ML to AI

new_string = string.replace("ML","AI/ML")

print(new_string)

```
