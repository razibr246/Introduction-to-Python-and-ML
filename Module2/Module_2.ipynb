C. Even, Odd, Positive and Negative


```python
n = int(input())

inp = input()

num = inp.split()

positive =0 ; negative =0 ; even =0 ; odd =0 


for i in range(n):
    x = int(num[i])
    # positive or negative
    if x > 0 :
        positive+=1
    elif x < 0 :
        negative+=1 
    
    # odd or even
    if x % 2 !=0 :
        odd+=1 
    else:
        even+=1 


print("Even:",even)
print("Odd:",odd)
print("Positive:",positive)
print("Negative:",negative)
```

### Q. Digits



```python
n = int(input())

for _ in range(n):
    num = input().strip()
    
    for digit in num[::-1]:   # reverse string
        print(digit, end=" ")
    
    print()
```
