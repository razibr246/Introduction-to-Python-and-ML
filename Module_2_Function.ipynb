### Different type of Argument

```python
def adder(a,b,c):

  print(f"{a},{b},{c}")


adder(3,6,2)
```

```python
def adder(*args):
  print(args)

adder(34,32,43,34,64)
```

```python
def adder(*args):
  for i in args:
    print(i)

adder(34,32,43,34,64)
```

```python
def adder(a,b,c):

  print(f"{a},{b},{c}")


adder(a =3,b =6,c = 2)
```

```python
def adder(**kwargs) :
  print(kwargs)

adder(a = 45,b = 43,c = 64,n_estimator = 99)
```

```python
def adder(**kwargs) :
  for key,value in kwargs.items():
    print(f"{key}:{value}")

adder(a = 45,b = 43,c = 64,n_estimator = 99)
```

### Iterator

```python
s = {10,12,23,34,4,556}

s_iter = iter(s)

print(s_iter)
print(next(s_iter))
print(next(s_iter))
print(next(s_iter))
print(next(s_iter))
print(next(s_iter))

for i in s:
  print(i)
```

```python
lst = [ x for x in range(500) ] # dataset

def data_loader(chunk_size,lst):
    for i in range(0,len(lst),chunk_size):
        yield lst[i:i+chunk_size]


x = data_loader(5,lst)

print(next(x))
print(next(x))
```
