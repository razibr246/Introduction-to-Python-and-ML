Operators IN Python

```python
x = 10 ; y = 3

sum = x + y 
sub = x - y
mult = x * y
div = round(x / y,3)

rem = x % y

print(sum,sub,mult,div,rem)
```

```python
x = 8; y = 9

greater_than = x > y

greater_than_equal = (2 >=2)

print(greater_than,greater_than_equal)


less_than = x < y

less_than_equal = (3<=3)

print(less_than,less_than_equal)




```
